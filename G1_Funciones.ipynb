Una red transmite datos a $100$ megabits por segundo. Crea una función utilizando *Python* que permita calcular la cantidad de datos transmitidos en una cierta cantidad de tiempo (en segundos). Luego utiliza la función y calcula cuántos datos se transmiten en:

* $45$ segundos
* $1,5$ minutos
* $1$ hora



```python
VELOCIDAD_DATOS = 100
def datos_totales(tiempo):
    total = VELOCIDAD_DATOS * tiempo
    return total

print(datos_totales(45)," megabits transmitidos en 45 segundos")
print(datos_totales(90)," megabits transmitidos en 90 segundos")
print(datos_totales(3600)," megabits transmitidos en 1 hora")
```

# Problema 3

La **latencia** de una red corresponde al tiempo que tarda un paquete de datos en viajar desde el punto de origen al destino. En otras palabras, es el tiempo que transcurre desde que se envía una solicitud de un dispositivo hasta que se recibe una respuesta del servidor u otro dispositivo.

La latencia puede verse afectada por la velocidad de transmisión de los datos a través de los cables o conexiones inalámbricas, la distancia física entre los dispositivos, el tiempo que tarda un computador en procesar o reenviar los datos, etc.

En redes de comunicación, como las videoconferencias, juegos en línea y transmisión de datos en tiempo real, es deseable una baja latencia, de lo contrario pueden existir retrasos en la comunicación y afectar negativamente la experiencia del usuario.

Según el contexto mencionado, crea una función en *Python* que permita calcular la **latencia real** de una red dada una latencia estimada. Suponga que la latencia real es un $20\%$ mayor a la latencia estimada.

Calcule la latencia real para una latencia estimada de:

a)   $200$ milisegundos.

b)   $149$ milisegundos.

c)   $74$ milisegundos.



```python
def latencia_real(latencia_estimada):
    real = round((latencia_estimada * 1.2),2)
    return real

print(latencia_real(200))
print(latencia_real(149))
print(latencia_real(74))
```

# Problema 4

Como parte de un proyecto de mejora, en el año $2021$ se instaló un cable de fibra óptica que une España con EEUU. El cable tiene un largo de $6.600$ $km$ y fue colocado por un barco a una velocidad de $1,85$ $km/h$.



1.   Determine la forma algebraica de la función que permite determinar el largo del cable instalado (en km) a partir del tiempo transcurrido (en horas). Utilice la instrucción *def* para definir la función en *Python*.
2.   Defina variable dependiente e independiente, indicando unidad de medida.
3.   Determine el dominio contextualizado de la función.
4.   Grafique la función utilizando la biblioteca *Matplotlib* considerando el dominio contextualizado.
5.   ¿Cuántos metros de cable se instalaron al transcurrir $148$ horas? ¿Y transcurridas $2.300$ horas?
6.   Si se han instalado $3.480$ kilometros de cable, ¿cuántas horas llevan de trabajo?
7.   ¿Cuánto tiempo transcurrió para que se completara la obra?




```python
#1.- Si velocidad se calcula como largo/tiempo, el largo se calculará como velocidad * tiempo, simplificando efectivamente las unidades de medida de "h"

VELOCIDAD = 1.85 #KM/H
def largo_cable(tiempo):
    return (VELOCIDAD * tiempo)

#2.- La variable independiente sería el tiempo transcurrido, pues puede crecer independientemente del largo del cable instalado. Por otro lado el largo que lleva el cable instalado se considera una variable dependiente, pues depende del tiempo transcurrido.
#3.- El dominio de la función correspondería a todos los reales mayores que 0 y menores que 3567,6, que es el número de horas que toma terminar la faena, pues no existe el tiempo negativo realisticamente en este contexto. por ende sería R (3567.6 >= tiempo >= 0) o bien
#DOM [0;3567,6]
#4.- 
tiempos= []
cableado = []
for i in range(0,100000):
    if (largo_cable(i)) <= 6600:
        tiempos.append(i)
        cableado.append(largo_cable(i))

plt.plot(tiempos,cableado)
plt.title("Relación entre largo del cable instalado y tiempo transcurrido")
plt.xlabel("Tiempo transcurrido (Horas)")
plt.ylabel("Largo de cable instalado (KM)")

plt.grid(True)
plt.show()

#5.- 
print(largo_cable(148), f"KMS de cable en 148 Horas")

#6.- Podemos redefinir una función para calcular tiempo, despejando la función inicial velocidad = largo/tiempo para aislar el tiempo, obteniendo que tiempo = largo / velocidad, luego:
def tiempo_utilizado(largo):
    return round((largo / VELOCIDAD),2)

#finalmente...
print(tiempo_utilizado(3480), f"horas para 3480 kilometros de cable")

#7.- Para que la obra esté completada, el cableado debería haber llegado a 6600 KMS, por lo tanto, reemplazamos 6600 en el argumento de la función.

print(tiempo_utilizado(6600), f"horas para completar la obra")




















```

# Problema 7 ( EJERCITACIÓN)

El número de usuarios de una red social se puede modelar mediante la función:

$$U(t) = \frac{1.000}{1+9e^{-0.5t}}$$

donde $t$ corresponde al tiempo transcurrido en meses.

1.  Defina variable dependiente e independiente, indicando unidad de medida.
2.  Determine la cantidad de usuarios transcurridos $12$ meses.
3.  Grafique la función, utilizando la librería *Matplotlib*, para los primeros dos años de funcionamiento.
4.  ¿Cuánto tiempo debe pasar para que la red social llegue a $800$ usuarios?

1) variable independiente t:tiempo transcurrido (meses)
variable dependiente U(t): Numero de usuarios de una red social (unidades)

```python
#2
from math import e
#def U(t):
  #return 1000 / (1+ 9* np.exp(-0.5*t)) Otro método para usar euler

def U(t):
  return 1000 / (1+ 9* (e**(-0.5*t)))

print(f"Transcurridos 12 meses, la red social tiene {U(12):.0f} usuarios aproximadamente")

```

```python
#3
t = np.arange(0,24,0.01)
plt.plot(t,U(t))
plt.title("Usuarios a lo largo de 2 años de la red social")
plt.xlabel("Tiempo transcurrido (meses)")
plt.ylabel("Cantidad de usuarios (unidades)")

plt.grid(True)
plt.show()
```

```python
#4
from scipy.optimize import fsolve

def tiempo(t):
    return U(t) - 800

t = np.linspace(0,10,10)
solucion = fsolve(tiempo,t)

print(f"La red social tendrá 800 usuarios transcurridos {solucion[0]:.2f} meses aproximadamente")
```

# Problema 8

El tiempo de ejecución de un algoritmo de ordenación (en segundos) se puede modelar mediante la función:

$$A(n) = 0,01n^{2} + 0,5n + 2$$

donde $n$ corresponde a la cantidad de elementos a ordenar. El algoritmo funciona desde las 23:00 horas hasta las 06:00 horas y debe ordenar $1.562$ elementos.

1. Defina variable dependiente e independiente, indicando unidad de medida.
2. Determine el dominio contextualizado de la función.
3. Determine el tiempo de ejecución para ordenar $1.200$ elementos.
4. Grafique, utilizando la biblioteca *Matplotlib*, la función $A(n)$.
5. ¿Cuántos elementos ordena luego de $6$ horas de funcionamiento?

```python
#1 La variable dependiente corresponde al tiempo de ejecución en segundos, mientras que la variable independiente corresponde a los elementos a ordenar, pues a mayor cantidad de datos, mayor tiempo se demorará en ordenar
#2 el dominio contextualizado de la función es dom A(n){0,1,2,....,1562}
#3
def A(n):
    return  0.01 * (n**2) + (0.5*n) + 2

print (f"Para ordenar 1200 datos se requieren {A(1200):.0f} segundos")

#4
n = np.arange(0,1562,0.0001)
plt.plot(n,A(n))
plt.title("Relación entre elementos a organizar y tiempo transcurrido")
plt.xlabel("Elementos a organizar")
plt.ylabel("Tiempo transcurrido")
plt.grid(True)
plt.show()

#5 & horas se transforman a segundos basandose en la siguiente formula:
def segundos(horas):
    return horas*3600
#Por lo tanto, 6 horas se registran como...
print(f"Seis horas corresponden a {segundos(6)}")

#Luego rehacemos el gráfico analizando solamente el rango en el que han transcurrido esa cantidad de segundos

segundos = np.arange(1444,1446,0.0001)
plt.plot(segundos,A(segundos))
plt.title("Relación entre elementos a organizar y tiempo transcurrido")
plt.xlabel("Elementos a organizar (unidades)")
plt.ylabel("Tiempo transcurrido (segundos)")
plt.grid(True)
plt.show()

#Por lo que se obtiene que se ordenan entre 1444 y 1445 elementos en ese rango de tiempo

#Alternativamente se puede realizar por FSOLVE

from scipy.optimize import fsolve

#Luego despejamos la función para que quede en 0, obteniendo 0 = 0.01n**2 + 0.5*n + 2 - 21600

def tiempo(n):
    return A(n) - 21600

n = np.linspace(0,50000,10)

solucion = fsolve(tiempo,n)
print(f"Se ordenan aproximadamente {solucion[0]:.0f} elementos en 6 horas de funcionamiento ")





```
